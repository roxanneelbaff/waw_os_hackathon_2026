
## Section 3 — Retrieval

Now that we have a local index (and access to remote ones) we can **retrieve** the most relevant passages for any query.

All three retrieval functions share an identical interface, so they are drop-in replacements:

```python
retrieve_*(queries, n=5)  →  list of result dicts
```

Each result dict contains:

| Key | Description |
|-----|-------------|
| `source` | Where the result came from: `local`, `remote`, or `ourss` |
| `score` | Relevance score — higher is more relevant |
| `text` | The retrieved passage |
| `url` | Original web page URL |
| `title` | Page title |

**Multi-query support:** pass a single string *or* a Python list of strings. For lists, results from all queries are merged and deduplicated by URL, ranked by score.

````python
# ── 4.1  retrieve_local — Local MOSAIC index 

from m1.retriever import retrieve_local
from m1.utils.mosaic_tools import display_results


local_hits = retrieve_local("open web search in europe", top=10)
display_results(local_hits)
````

````python
# ── 4.2  retrieve_remote — Remote MOSAIC index ────────────────────────────────
# Add REMOTE_MOSAIC_URL to your .env file once the organizers provide it.
from m1.retriever import retrieve_remote, display_results


remote_hits = retrieve_remote("earth", top=5)
display_results(remote_hits, "Remote MOSAIC — test query")
````

````python
# ── 4.3  retrieve_ourss — ourrs.eu Open Search API 
from m1.retriever import retrieve_ourss


ourss_hits = retrieve_ourss(["open web search"], top=5)
display_results(ourss_hits, "ourrs.eu — test query")
````
