
## Section 2 — Indexing with MOSAIC

### What is MOSAIC?

**MOSAIC** is an open-source local search engine that runs entirely on your laptop — no account, no cloud, no fees. It is built on **Apache Lucene** (the same engine that powers enterprise search products like Elasticsearch) and can return search results in milliseconds.

Think of it as your own personal Google Search for the OWI data you downloaded.

> MOSAIC homepage: https://mosaic.ows.eu

### What you need before starting

MOSAIC is a **Java** application. You need:
- **Java 17 or newer** — use an open-source distribution (Temurin, Corretto, Eclipse Adoptium). **Oracle JDK is not allowed on company laptops.**
- **Apache Maven 3.8+** — the build tool
- **Git** — to download the MOSAIC source code

The cells below check all prerequisites automatically.

### Installing Java 17 and Maven

MOSAIC requires **Java 17** (Temurin, free & open-source) and **Maven**. Install both before running the cells below.

---

#### macOS
```bash
brew install --cask temurin@17
brew install maven
```

---

#### Windows
**Java 17** — download and run the `.msi` installer from:  
https://adoptium.net/temurin/releases/?version=17

**Maven** — easiest via winget:
```cmd
winget install Apache.Maven
```
Or manually: download the zip from https://maven.apache.org/download.cgi, unzip it, then add its `bin/` folder to your PATH:
```
MAVEN_HOME = C:\apache-maven-3.x.x
PATH       = %PATH%;%MAVEN_HOME%\bin
```

---

#### Linux — Ubuntu / Debian
```bash
# Add the Adoptium repo (one-time)
wget -qO - https://packages.adoptium.net/artifactory/api/gpg/key/public | sudo apt-key add -
echo "deb https://packages.adoptium.net/artifactory/deb $(lsb_release -cs) main" | sudo tee /etc/apt/sources.list.d/adoptium.list

sudo apt update
sudo apt install temurin-17-jdk maven
```

---

#### Linux — RHEL / Fedora
```bash
# Add the Adoptium repo (one-time)
sudo rpm --import https://packages.adoptium.net/artifactory/api/gpg/key/public
# create /etc/yum.repos.d/adoptium.repo — see adoptium.net for the full repo file

sudo dnf install temurin-17-jdk maven
```

---

After installing, verify both in a **new terminal**:
```bash
java -version   # should show openjdk version "17.x.x"
mvn --version   # should show Apache Maven 3.x.x
```
